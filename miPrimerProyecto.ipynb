```python
def calculorSueldo(hrs, valor):
  salario_bruto = float(hrs) * float(valor) * 30
  isss = float(salario_bruto * 0.03)  # ISSS es el 3%

  if isss >= 30:
    isss = 30.0

  afp = float(salario_bruto * 0.0725)  # AFP es el 7.25%

  # Renta imponible = salario bruto menos ISSS y AFP
  renta_imponible = salario_bruto - (isss + afp)

  # El ISR se calcula sobre la renta imponible, no sobre el salario bruto
  isr = calcularIsr(renta_imponible)

  sueldo_final = renta_imponible - isr

  print(f"El salario bruto es: {salario_bruto}")
  print(f"El isss es: {isss}")
  print(f"El afp es: {afp}")
  print(f"La renta imponible es: {round(renta_imponible,2)}")
  print(f"El isr retenido es: {round(isr,2)}")
  print(f"El sueldo neto final es: {round(sueldo_final,2)}")


def calcularIsr(monto):
  monto = float(monto)
  # Calculamos el impuesto sobre renta (fórmula oficial: excedente + cuota fija)

  # Tramo 1
  if monto >= 0.01 and monto <= 472.00:
    return 0
  # Tramo 2
  elif monto >= 472.01 and monto <= 895.24:
    return (monto - 472.00) * 0.10 + 17.67
  #Tramo 3
  elif monto >= 895.25 and monto <= 2038.10:
    return (monto - 895.24) * 0.20 + 60.00
  else:
    # Tramo 4
    return (monto - 2038.10) * 0.30 + 288.57
```

```python
from functools import reduce
import json

class Estadistica:
    # media aritmetica
    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - media) ** 2, serie, 0)
        return suma / len(serie)

    # Desviacion estandar y tipica
    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def media_armonica(self, datos):
        n = len(datos)
        suma_inversos = sum(1 / x for x in datos)
        return n / suma_inversos

    def frecuencia(self, datos):
        frecuencias = {}
        for dato in datos:
            if dato in frecuencias:
                frecuencias[dato] += 1
            else:
                frecuencias[dato] = 1
        return frecuencias

    def palindromo(self, palabra):
        palabra = palabra.lower()
        palabra = palabra.replace(" ", "")
        palabra_invertida = palabra[::-1]
        return palabra == palabra_invertida

    # moda
    def moda(self, serie):
        frecuencia = self.frecuencia(serie)
        moda = max(frecuencia, key=frecuencia.get)
        return moda

    # Rango
    def rango(self, serie):
        menos = min(serie)
        mas = max(serie)
        return mas - menos
```

Sistema Academico usando JSON/diccionarios

Dataset: 20 datos reales obtenido de sus compañeros: Nombres, Materias, notas, Edades, tiempo de traslado a la UGB, horas de uso del telefono.

Usando:

POO, funciones de usuario y predefinidas. Estructuras de control, listas, diccionarios, etc.

```python
class Estudiante:
    def __init__(self, nombre, edad, tiempo_traslado_min, horas_telefono, materias):
        self.nombre = nombre
        self.edad = edad
        self.tiempo_traslado_min = tiempo_traslado_min
        self.horas_telefono = horas_telefono
        self.materias = materias

    # Obtener notas por materia que tiene el estudiante
    def notas_materia(self, materia):
        # Usando filter para encontrar la materia
        materia_encontrada = list(filter(
            lambda m: m["materia"] == materia,
            self.materias
        ))

        if materia_encontrada:
            # Usando map para extraer las notas
            return list(map(
                lambda clave: materia_encontrada[0][clave],
                ["lab1", "lab2", "parcial"]
            ))
        return None

    def promedio_general(self):
        promedios_materias = list(map(
            lambda m: m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40,
            self.materias
        ))
        return sum(promedios_materias) / len(promedios_materias)

    def aprobo(self, nota_minima=6.0):
        promedio = self.promedio_general()
        return promedio >= nota_minima

    def materias_aprobadas(self, nota_minima=6.0):
        return list(filter(
            lambda m: (m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40) >= nota_minima,
            self.materias
        ))


    def promedio_por_materia(self):
        return list(map(
            lambda m: {
                "materia": m["materia"],
                "promedio": m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40
            },
            self.materias
        ))

    def to_dict(self):
        """Convierte el estudiante a diccionario"""
        return {
            "nombre": self.nombre,
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado_min,
            "horas_telefono": self.horas_telefono,
            "materias": self.materias,
            "promedio_general": self.promedio_general(),
            "aprobado": self.aprobo()
        }

class SistemaAcademico:
    def __init__(self):
        self.estudiantes = []
        self.estadistica = Estadistica()

    def cargar_estudiantes(self, datos):
        """Carga estudiantes desde un dataset de diccionarios usando map"""
        self.estudiantes = list(map(
            lambda dato: Estudiante(
                dato["nombre"],
                dato["edad"],
                dato["tiempo_traslado_min"],
                dato["horas_telefono"],
                dato["materias"]
            ),
            datos
        ))

    def reporte_general(self):
        """Reporte general de edad, traslado y telefono usando map y filter"""
        # Usando map para extraer datos
        edades = list(map(lambda e: e.edad, self.estudiantes))
        tiempos_traslado = list(map(lambda e: e.tiempo_traslado_min, self.estudiantes))
        horas_telefono = list(map(lambda e: e.horas_telefono, self.estudiantes))

        print("\n=== REPORTE GENERAL ===")
        print(f"Edad - Promedio: {self.estadistica.media(edades):.2f}, Desviacion: {self.estadistica.desviacion(edades):.2f}")
        print(f"Tiempo traslado - Promedio: {self.estadistica.media(tiempos_traslado):.2f}, Desviacion: {self.estadistica.desviacion(tiempos_traslado):.2f}")
        print(f"Horas telefono - Promedio: {self.estadistica.media(horas_telefono):.2f}, Desviacion: {self.estadistica.desviacion(horas_telefono):.2f}")

    def reporte_campo(self, campo):
        """Reporte de un campo especifico usando map y filter"""
        # Usando map para extraer valores
        valores = list(map(
            lambda e: getattr(e, campo, None),
            self.estudiantes
        ))
        # Filtrar valores None
        valores = list(filter(lambda x: x is not None, valores))

        if valores:
            print(f"\n=== REPORTE DE {campo.upper()} ===")
            print(f"Media: {self.estadistica.media(valores):.2f}")
            print(f"Desviacion: {self.estadistica.desviacion(valores):.2f}")
            print(f"Moda: {self.estadistica.moda(valores)}")
            print(f"Rango: {self.estadistica.rango(valores)}")
            print(f"Frecuencia: {self.estadistica.frecuencia(valores)}")
        else:
            print("Campo no encontrado")

    def reporte_componente(self, componente):
        """Reporte de un componente de notas (lab1, lab2, parcial) usando map y filter"""
        # Usando map y filter para extraer todas las notas de ese componente
        notas = list(filter(
            lambda x: x is not None,
            list(map(
                lambda e: list(map(
                    lambda m: m.get(componente, None),
                    e.materias
                )),
                self.estudiantes
            ))
        ))
        # Aplanar la lista usando reduce
        notas_planas = reduce(lambda x, y: x + y, notas, [])
        # Filtrar None
        notas_limpias = list(filter(lambda x: x is not None, notas_planas))

        if notas_limpias:
            print(f"\n=== REPORTE DE {componente.upper()} ===")
            print(f"Media: {self.estadistica.media(notas_limpias):.2f}")
            print(f"Desviacion: {self.estadistica.desviacion(notas_limpias):.2f}")
            print(f"Moda: {self.estadistica.moda(notas_limpias)}")
            print(f"Rango: {self.estadistica.rango(notas_limpias)}")
        else:
            print("Componente no encontrado")

    def notas_por_materia(self):
        """Promedio de notas por materia usando map, filter y reduce"""
        # Obtener todas las materias únicas
        todas_materias = list(map(
            lambda e: [m["materia"] for m in e.materias],
            self.estudiantes
        ))
        materias_unicas = list(set(reduce(lambda x, y: x + y, todas_materias, [])))

        # Calcular promedios por materia
        promedios_por_materia = {}
        for materia in materias_unicas:
            # Filtrar estudiantes que tienen esta materia
            notas_materia = list(filter(
                lambda x: x is not None,
                list(map(
                    lambda e: e.notas_materia(materia),
                    self.estudiantes
                ))
            ))
            # Calcular promedio de todas las notas de esta materia
            todas_notas = reduce(lambda x, y: x + y, notas_materia, [])
            if todas_notas:
                promedio = sum(todas_notas) / len(todas_notas)
                promedios_por_materia[materia] = promedio

        print("\n=== PROMEDIO DE NOTAS POR MATERIA ===")
        for materia, promedio in promedios_por_materia.items():
            print(f"{materia}: {promedio:.2f}")

    def estudiantes_aprobados(self, nota_minima=6.0):
        """Ver estudiantes aprobados usando filter"""
        return list(map(
            lambda e: (e.nombre, e.promedio_general()),
            filter(lambda e: e.aprobo(nota_minima), self.estudiantes)
        ))

    def estudiantes_reprobados(self, nota_minima=6.0):
        """Ver estudiantes reprobados usando filter"""
        return list(map(
            lambda e: (e.nombre, e.promedio_general()),
            filter(lambda e: not e.aprobo(nota_minima), self.estudiantes)
        ))

    def resumen_por_materia(self):
        """Resumen estadistico completo por materia usando map y reduce"""
        # Obtener todas las materias únicas
        todas_materias = list(map(
            lambda e: [m["materia"] for m in e.materias],
            self.estudiantes
        ))
        materias_unicas = list(set(reduce(lambda x, y: x + y, todas_materias, [])))

        print("\n=== RESUMEN ESTADISTICO POR MATERIA ===")
        for materia in materias_unicas:
            # Obtener todas las notas de esta materia
            notas = list(map(
                lambda e: list(filter(
                    lambda x: x is not None,
                    e.notas_materia(materia) or []
                )),
                self.estudiantes
            ))
            # Aplanar la lista
            notas_planas = reduce(lambda x, y: x + y, notas, [])

            if notas_planas:
                print(f"\n{materia}:")
                print(f"  Media: {self.estadistica.media(notas_planas):.2f}")
                print(f"  Desviacion: {self.estadistica.desviacion(notas_planas):.2f}")
                print(f"  Moda: {self.estadistica.moda(notas_planas)}")
                print(f"  Rango: {self.estadistica.rango(notas_planas)}")


    def mostrar_estudiantes(self):
        """Muestra todos los estudiantes como diccionarios completos"""
        print("\n=== LISTA DE ESTUDIANTES ===")
        print("="*80)

        # Usar map para convertir cada estudiante a diccionario
        estudiantes_dict = list(map(lambda e: e.to_dict(), self.estudiantes))

        # Mostrar cada estudiante como JSON formateado
        for i, estudiante in enumerate(estudiantes_dict, 1):
            print(json.dumps(estudiante, ensure_ascii=False, indent=2))
            print("-"*80)

    def estudiantes_mayores_edad(self, edad_minima=18):
        """Filtra estudiantes mayores de cierta edad usando filter"""
        return list(filter(lambda e: e.edad >= edad_minima, self.estudiantes))

    def promedio_general_grupo(self):
        """Promedio general de todos los estudiantes usando reduce"""
        promedios = list(map(lambda e: (e.nombre, e.promedio_general()), self.estudiantes))

        print("\n=== PROMEDIO GENERAL DEL GRUPO ===")
        for nombre, promedio in promedios:
            print(f"  {nombre}: {promedio:.2f}")

        solo_promedios = list(map(lambda p: p[1], promedios))
        promedio_grupo = reduce(lambda x, y: x + y, solo_promedios, 0) / len(solo_promedios) if solo_promedios else 0

        return promedio_grupo

    def ver_aprobados_reprobados(self, nota_minima=6.0):
        """Muestra estudiantes aprobados y reprobados"""
        aprobados = self.estudiantes_aprobados(nota_minima)
        reprobados = self.estudiantes_reprobados(nota_minima)

        print("\n" + "="*50)
        print("RESUMEN DE APROBADOS Y REPROBADOS")
        print("="*50)

        # Aprobados
        print(f"\nAPROBADOS ({len(aprobados)} estudiantes):")
        if aprobados:
            for nombre, promedio in aprobados:
                print(f"  - {nombre}: {promedio:.2f}")
        else:
            print("  Ningun estudiante aprobo")

        # Reprobados
        print(f"\nREPROBADOS ({len(reprobados)} estudiantes):")
        if reprobados:
            for nombre, promedio in reprobados:
                print(f"  - {nombre}: {promedio:.2f}")
        else:
            print("  Ningun estudiante reprobo")

        # Estadisticas adicionales
        total = len(self.estudiantes)
        if total > 0:
            tasa_aprobacion = (len(aprobados) / total) * 100
            print(f"\nTasa de aprobacion: {tasa_aprobacion:.1f}%")

    def promedios_por_estudiante(self):
        """Muestra el promedio por materia y el promedio general de cada estudiante"""
        print("\n=== PROMEDIO GENERAL POR ESTUDIANTE ===")
        for e in self.estudiantes:
            print(f"\n{e.nombre}")
            print("-" * len(e.nombre))
            for resultado in e.promedio_por_materia():
                print(f"  {resultado['materia']:<35}: {resultado['promedio']:.2f}")
            print(f"  {'Promedio general':<35}: {e.promedio_general():.2f}")

# DATASET: estudiantes
# DATASET: estudiantes
datos_estudiantes = [
    {"nombre": "Jimmy Roberto Martinez Gomez", "edad": 19,
        "tiempo_traslado_min": 30, "horas_telefono": 6,
        "materias": [
            {"materia": "Administracion de Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion Computacional III", "lab1": 9.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.0, "lab2": 8.3, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10, "lab2": 10, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10, "lab2": 10, "parcial": 0},
        ]
    },
    {"nombre": "Wilfredo Josue Ortez Gaitan", "edad": 20,
        "tiempo_traslado_min": 50, "horas_telefono": 6,
        "materias": [
            {"materia": "Administracion de Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion Computacional III", "lab1": 9, "lab2": 9, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8, "lab2": 8.3, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10, "lab2": 10, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10, "lab2": 10, "parcial": 0},
        ]
    },
    {"nombre": "Miguel Alejandro Gómez Saravia", "edad": 19,
        "tiempo_traslado_min": 15, "horas_telefono": 4,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Josue Stiven Funes Benavides", "edad": 19,
        "tiempo_traslado_min": 30, "horas_telefono": 5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Jefferson Steven Castillo Leiva", "edad": 19,
        "tiempo_traslado_min": 15, "horas_telefono": 2.3,  # 2h 18min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Ramael Ernesto Cordova Colindres", "edad": 22,
        "tiempo_traslado_min": 7, "horas_telefono": 13.52,  # 13h 31min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Carlos Daniel Parada Avalos", "edad": 25,
        "tiempo_traslado_min": 10, "horas_telefono": 5.52,  # interpretado como 5h 31min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
        ]
    },
    {"nombre": "Yeimy Julissa Parada Chavez", "edad": 20,
        "tiempo_traslado_min": 45, "horas_telefono": 3,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Madison Odalis Valle Juarez", "edad": 19,  # no se proporciono
        "tiempo_traslado_min": 30, "horas_telefono": 9,
        "materias": [  # asumido mismo set de materias del grupo; solo dijo "todas son 10"
            {"materia": "Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Jennifer Maricela Renderos Guevara", "edad": 21,
        "tiempo_traslado_min": 60, "horas_telefono": 3,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 8.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Kimberly Michelle Castro Zelaya", "edad": 18,
        "tiempo_traslado_min": 20, "horas_telefono": 9,
        "materias": [  # solo se proporciono esta materia
            {"materia": "Programacion Computacional III", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Samuel Alexander Portillo Ceron", "edad": 20,
        "tiempo_traslado_min": 80,  # 1:20 interpretado como 1h20min
        "horas_telefono": 6,
        "materias": [
            {"materia": "Estadistica", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Base de Datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },



    # DATOS PARA LLENAR LOS 20

        {"nombre": "Andrea Beatriz Chavez Rivas", "edad": 20,
        "tiempo_traslado_min": 25, "horas_telefono": 4.5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 7.0, "lab2": 7.5, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 8.0, "lab2": 8.5, "parcial": 0},
            {"materia": "Programacion III", "lab1": 6.5, "lab2": 7.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 7.5, "lab2": 8.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 7.0, "lab2": 7.0, "parcial": 0},
        ]
    },
    {"nombre": "Oscar Ivan Mejia Portillo", "edad": 23,
        "tiempo_traslado_min": 40, "horas_telefono": 7,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 6.0, "lab2": 6.5, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 7.0, "lab2": 7.5, "parcial": 0},
            {"materia": "Programacion III", "lab1": 5.5, "lab2": 6.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 6.5, "lab2": 7.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 6.0, "lab2": 6.5, "parcial": 0},
        ]
    },
    {"nombre": "Sofia Nicole Aguilar Reyes", "edad": 18,
        "tiempo_traslado_min": 10, "horas_telefono": 3,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.5, "lab2": 9.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 9.5, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 9.5, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.5, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 9.0, "parcial": 0},
        ]
    },
    {"nombre": "Luis Fernando Escobar Molina", "edad": 21,
        "tiempo_traslado_min": 35, "horas_telefono": 6.5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 7.5, "lab2": 8.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 6.0, "lab2": 6.5, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 8.5, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 7.0, "lab2": 7.5, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.5, "lab2": 8.0, "parcial": 0},
        ]
    },
    {"nombre": "Karen Michelle Guardado Lopez", "edad": 19,
        "tiempo_traslado_min": 20, "horas_telefono": 5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 8.5, "lab2": 9.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 7.5, "lab2": 8.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 8.0, "lab2": 8.5, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.5, "lab2": 9.0, "parcial": 0},
        ]
    },
    {"nombre": "Bryan Alexander Hernandez Cruz", "edad": 24,
        "tiempo_traslado_min": 55, "horas_telefono": 8,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 5.0, "lab2": 5.5, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 6.0, "lab2": 6.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 5.5, "lab2": 6.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 6.0, "lab2": 6.5, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 5.0, "lab2": 5.5, "parcial": 0},
        ]
    },
    {"nombre": "Daniela Alejandra Ventura Sosa", "edad": 20,
        "tiempo_traslado_min": 18, "horas_telefono": 4,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 9.5, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 8.5, "lab2": 9.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.5, "lab2": 9.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 9.5, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.5, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Marvin Ernesto Alvarado Diaz", "edad": 22,
        "tiempo_traslado_min": 12, "horas_telefono": 2.5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 7.0, "lab2": 6.5, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 7.5, "lab2": 7.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 8.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 6.5, "lab2": 7.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 7.0, "lab2": 7.5, "parcial": 0},
        ]
    },
]

# MENU PRINCIPAL
def menu():
    sistema = SistemaAcademico()
    sistema.cargar_estudiantes(datos_estudiantes)

    while True:
        print("\n========== SISTEMA ESTADISTICO ACADEMICO ==========")
        print("1. Reporte general (edad, traslado, telefono)")
        print("2. Reporte de un campo especifico")
        print("3. Promedio de notas por materia")
        print("4. Ver aprobados y reprobados")
        print("5. Ver todos los estudiantes")
        print("6. Resumen estadistico completo por materia")
        print("7. Promedio general por estudiante")
        print("8. Promedio general del grupo")
        print("0. Salir")

        opcion = input("Elige una opcion: ")

        if opcion == "1":
            sistema.reporte_general()

        elif opcion == "2":
            print("Campos disponibles:\n1-edad\n2-tiempo traslado\n3-horas telefono\n4-lab1\n5-lab2\n6-parcial")
            campo = input("Escribe el numero del campo: ")

            if campo == "1":
                sistema.reporte_campo("edad")
            elif campo == "2":
                sistema.reporte_campo("tiempo_traslado_min")
            elif campo == "3":
                sistema.reporte_campo("horas_telefono")
            elif campo == "4":
                sistema.reporte_componente("lab1")
            elif campo == "5":
                sistema.reporte_componente("lab2")
            elif campo == "6":
                sistema.reporte_componente("parcial")
            else:
                print("Campo no valido.")

        elif opcion == "3":
            sistema.notas_por_materia()

        elif opcion == "4":
            sistema.ver_aprobados_reprobados()

        elif opcion == "5":
            sistema.mostrar_estudiantes()

        elif opcion == "6":
            sistema.resumen_por_materia()

        elif opcion == "7":
            sistema.promedios_por_estudiante()

        elif opcion == "8":
            promedio_grupo = sistema.promedio_general_grupo()
            print(f"\nPromedio general del grupo: {promedio_grupo:.2f}")


        elif opcion == "0":
            print("Saliendo del sistema...")
            break

        else:
            print("Opcion invalida, intenta de nuevo.")

# Correr el sistema
menu()
```
