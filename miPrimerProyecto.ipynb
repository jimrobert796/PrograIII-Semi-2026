```python
def calculorSueldo(hrs, valor):
  salario_bruto = float(hrs) * float(valor) * 30
  isss = float(salario_bruto * 0.03)  # ISSS es el 3%

  if isss >= 30:
    isss = 30.0

  afp = float(salario_bruto * 0.0725)  # AFP es el 7.25%

  # Renta imponible = salario bruto menos ISSS y AFP
  renta_imponible = salario_bruto - (isss + afp)

  # El ISR se calcula sobre la renta imponible, no sobre el salario bruto
  isr = calcularIsr(renta_imponible)

  sueldo_final = renta_imponible - isr

  print(f"El salario bruto es: {salario_bruto}")
  print(f"El isss es: {isss}")
  print(f"El afp es: {afp}")
  print(f"La renta imponible es: {round(renta_imponible,2)}")
  print(f"El isr retenido es: {round(isr,2)}")
  print(f"El sueldo neto final es: {round(sueldo_final,2)}")


def calcularIsr(monto):
  monto = float(monto)
  # Calculamos el impuesto sobre renta (fórmula oficial: excedente + cuota fija)

  # Tramo 1
  if monto >= 0.01 and monto <= 472.00:
    return 0
  # Tramo 2
  elif monto >= 472.01 and monto <= 895.24:
    return (monto - 472.00) * 0.10 + 17.67
  #Tramo 3
  elif monto >= 895.25 and monto <= 2038.10:
    return (monto - 895.24) * 0.20 + 60.00
  else:
    # Tramo 4
    return (monto - 2038.10) * 0.30 + 288.57
```

```python
horas = input("Ingrese el las horas trabajadas: ")
valor = input("Ingrese el pago por hora: ")

calculorSueldo(horas, valor)
```

**Funciones preconstruidas**

```python
# Funcion o metodo .map()
# Map funciona para en una lista [] se quiera hacer multiples cosas con una funcion

semana = ["Domingo", "lunes", "martes", "miercoles", "jueves", "Viernes", "Sabado"]
print(semana[0])

primerLetraDia = map(lambda dia: dia[0].lower(), semana) # Map nos devuelve un objeto el cual lo podemos usar en lista
print(list(primerLetraDia))

primerLetraDia = map(lambda dia: dia[0].upper(), semana)
print(list(primerLetraDia))
```

```python
#Todos los numeros de la matriz elevarlos a un exponente
numeros = [2,3,4,5]

exponente = int(input("Ingrese el exponente: "))

conversion = map(lambda num:
                 num ** exponente,
                 numeros)
print(list(conversion))

```

```python
# Funcion filter
# Dada una serie de numeros necesitamos los unicos que sean pares
serie = [1,2,3,4,5,6,7,8,9]

pares = filter(lambda num:
               num % 2 == 0,
               serie
               )
impares = filter(lambda num:
               num % 2 != 0,
               serie
               )
print("Impares: ", list(impares))
print("Pares: ", list(pares))
```

```python
#Clasificar la nota de aprovados y reprovados

notas = [2,3,4,5,6,7,8,4,6,7,6,7,5]

aprovados = filter(lambda nota:
                   nota >= 6,
                   notas
                   )

reprovados = filter(lambda nota:
                   nota < 6,
                   notas
                   )

print("Aprovados:", list(aprovados))
print("Reprovados:", list(reprovados))
```

```python
#Cuanto es el acomulado de numeros pares de una serie
from functools import reduce

serie = [1,2,3,4,5,6,7,8,9]

pares = filter(lambda num:
               num % 2 == 0,
               serie
               )


sumaPar = reduce(lambda acum, num: num + acum, pares)

print("La suma de pares es:",sumaPar)

```

```python
import datetime
fecha = datetime.date.today()

dia = fecha.day
mes = fecha.month
year = fecha.year

print(fecha, "Dia: ", dia, "Mes: ", mes, "Año: ", year)

semana = ("domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado")
meses = ("", "Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto",
         "Septiembre", "Octubre", "Noviembre", "Diciembre")

print(semana, "Primer dia de la semana: ", semana[0])

diaSemana = semana[fecha.isoweekday()]
print(diaSemana, dia, "de", meses[mes], "del año", year)
```
