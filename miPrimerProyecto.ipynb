**Funciones en python**

```python
def resta(num1, num2):
  return float(num1) - float(num2)
```

```python
def suma(num1, num2):
  return float(num1) + float(num2)
```

```python
num1 = input("Ingrese el primer numero: ")
num2 = input("Ingrese el segundo numero: ")
print("La suma es: ", suma(num1, num2))
print("La resta es: ", resta(num1, num2))
```

Hacer en python el siguiente ejercicio haciendo uso de funciones, que convierta de grados celcuis a farenheit

```python
def convertir(celcuis):
  return float(celcuis) * 9/5 + 32
```

```python
celcuis = input("Ingrese los grados celcuis: ")
print("Los grados farenheit son: ", convertir(celcuis))
```

Hacer en python el siguiente ejercicio haciendo uso de funciones, que convierta de grados farenheit a celcuis

```python
def convertirFaren(farenheit):
  return (float(farenheit) -32) * 9/5
```

```python
faren = input("Ingrese los grados farenheit: ")
print("Los grados celcuis son: ", convertirFaren(faren))
```

Hacer en python el siguiente ejercicio haciendo uso de funciones, que convierta de grados celcuis a kelvin

```python
def convertirKel(celcuis):
  return float(celcuis) + 273.15
```

```python
cel = input("Ingrese los grados celcuis: ")
print("Los grados en kelvin son:", convertirKel(cel))
```

Hacer en python el siguiente ejercicio haciendo uso de funciones anonimas, que convierta de grados celcuis a farenheit

```python
convertir = lambda celcuis: float(celcuis) * 9/5 + 32
cel = input("Ingrese los grados celcuis: ")
print("Los grados en farenheit son:", convertir(cel))
```

Hacer en python el siguiente ejercicio haciendo uso de funciones anonimas:

- convertir de farenheit a celcuis
- convertir de kelvin a celcuis
- calcular el sueldo de un empleado. hrs * valor de horas * 30 dias
  las deduciones: ISSS, AFP, ISR
- elevar un numero a su exponente

```python
convertirF = lambda farenheit: (float(farenheit) -32) * 9/5
faren = input("Ingrese los grados farenheit: ")
print("Los grados celcuis son: ", convertirF(faren))
```

```python
convertirK = lambda kelvin: float(kelvin) + 273.15
kel = input("Ingrese los grados kelvin: ")
print("Los grados celcuis son: ", convertirK(kel))
```
