```python
def calculorSueldo(hrs, valor):
  salario_bruto = float(hrs) * float(valor) * 30
  isss = float(salario_bruto * 0.03)  # ISSS es el 3%

  if isss >= 30:
    isss = 30.0

  afp = float(salario_bruto * 0.0725)  # AFP es el 7.25%

  # Renta imponible = salario bruto menos ISSS y AFP
  renta_imponible = salario_bruto - (isss + afp)

  # El ISR se calcula sobre la renta imponible, no sobre el salario bruto
  isr = calcularIsr(renta_imponible)

  sueldo_final = renta_imponible - isr

  print(f"El salario bruto es: {salario_bruto}")
  print(f"El isss es: {isss}")
  print(f"El afp es: {afp}")
  print(f"La renta imponible es: {round(renta_imponible,2)}")
  print(f"El isr retenido es: {round(isr,2)}")
  print(f"El sueldo neto final es: {round(sueldo_final,2)}")


def calcularIsr(monto):
  monto = float(monto)
  # Calculamos el impuesto sobre renta (fórmula oficial: excedente + cuota fija)

  # Tramo 1
  if monto >= 0.01 and monto <= 472.00:
    return 0
  # Tramo 2
  elif monto >= 472.01 and monto <= 895.24:
    return (monto - 472.00) * 0.10 + 17.67
  #Tramo 3
  elif monto >= 895.25 and monto <= 2038.10:
    return (monto - 895.24) * 0.20 + 60.00
  else:
    # Tramo 4
    return (monto - 2038.10) * 0.30 + 288.57
```

```python
horas = input("Ingrese el las horas trabajadas: ")
valor = input("Ingrese el pago por hora: ")

calculorSueldo(horas, valor)
```

```python
# Funcion filter
# Dada una serie de numeros necesitamos los unicos que sean pares
serie = [1,2,3,4,5,6,7,8,9]

pares = filter(lambda num:
               num % 2 == 0,
               serie
               )
impares = filter(lambda num:
               num % 2 != 0,
               serie
               )
print("Impares: ", list(impares))
print("Pares: ", list(pares))
```

```python
# Haciendo uso de estructuras de control en Python, determinar si el número introducido es par/impar y primo

# Obtenemos datos
num = int(input("Ingrese un numero: "))

# Verificamos si es par o impar
if num % 2 == 0:
    var = "es par"
else:
    var = "es impar"

# Verificamos si es primo
es_primo = True

if num < 2:
    es_primo = False
else:
    for i in range(2, num):
        if num % i == 0:
            es_primo = False
            break  # Terminar el bucle, ya encontramos un divisor

# Mostramos el resultado una sola vez, al final
if es_primo:
    print(num, "es primo y", var)
else:
    print(num, "no es primo y", var)
```

```python
from functools import reduce
class estadistica:


  # media arimetica
  def media(self, serie):
    suma = reduce(lambda acum, num: acum+ float(num), serie, 0)
    return suma/len(serie)

  def varianza(self, serie):
    media = self.media(serie)
    #4 + 0 + 4 + 0 + 0 + 16= 24/6 = 4.0
    suma = reduce(lambda acum, num: acum + (float(num)-media)**2, serie, 0)
    return suma/len(serie)

  # Desviacion estandar y tipica
  def desviacion(self, serie):
    return self.varianza(serie)**0.5

  def media_armonica(self, datos):
      n = len(datos)
      suma_inversos = sum(1 / x for x in datos)
      return n / suma_inversos

  def frecuencia(self, datos):
      frecuencias = {}

      for dato in datos:
          if dato in frecuencias:
              frecuencias[dato] += 1
          else:
              frecuencias[dato] = 1
      return frecuencias

  def palindromo(self, palabra):
    palabra = palabra.lower()
    palabra = palabra.replace(" ", "")
    palabra_invertida = palabra[::-1]

    if palabra == palabra_invertida:
        return True
    else:
        return False

  # moda
  def moda(self, serie):
    frecuencia = self.frecuencia(serie)
    moda = max(frecuencia, key=frecuencia.get)
    return moda

  # Rango
  def rango(self, serie):
    menos = min(serie)
    mas = max(serie)
    return mas - menos
```

```python

#Hacer un programa en python que lea una serie de numeros introducida por el usuario
#y que calcule la media aritmetica
objEstadistica = estadistica()
serie = input("Serie: ").split(",")
print("Media: ", objEstadistica.media(serie))
print("Varianza: ", objEstadistica.varianza(serie))
print("Desviacion: ", objEstadistica.desviacion(serie))
```
