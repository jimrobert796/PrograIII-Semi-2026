```python
print("Sume dos numeros")
num1 = input("Ingrese numero: ")
num2 = input("Ingrese numero: ")


res = float(num1) + float(num2)
print(type(res))

print(f"Su resultado {res}")

res = "suma completada"
```

```python
print("Reste dos numeros")
num1 = input("Ingrese numero: ")
num2 = input("Ingrese numero: ")


res = float(num1) - float(num2)
print(type(res))

print(f"Su resultado {res}")

res = "suma completada"
```

```python
print("Multiplique dos numeros")
num1 = input("Ingrese numero: ")
num2 = input("Ingrese numero: ")


res = float(num1) * float(num2)
print(type(res))

print(f"Su resultado {res}")

res = "suma completada"
```

```python
print("divida dos numeros")
num1 = input("Ingrese numero: ")
num2 = input("Ingrese numero: ")


res = float(num1) / float(num2)
print(type(res))

print(f"Su resultado {res}")

res = "suma completada"
```
